```python
# ============================================================
# 0. Library
# ============================================================

import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

# 한글/마이너스 부호 관련 설정
plt.rcParams["axes.unicode_minus"] = False


# ============================================================
# 1. Load Dataset
# ============================================================

DATA_PATH = "dataset/winequality-red.csv"

df = pd.read_csv(DATA_PATH)

print("=" * 60)
print("DATASET")
print("=" * 60)

print(f"Shape: {df.shape}")

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 2. Basic Information
# ============================================================

print("\n" + "=" * 60)
print("BASIC INFORMATION")
print("=" * 60)

df.info()

print("\nData Types:")
print(df.dtypes)


# ============================================================
# 3. Missing Value Analysis
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUE ANALYSIS")
print("=" * 60)

missing_count = df.isnull().sum()
missing_ratio = df.isnull().mean() * 100

missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Ratio (%)": missing_ratio
})

display(missing_table)


# Missing value 위치
missing_positions = np.argwhere(df.isnull().values)

print("\nMissing Value Positions:")

if len(missing_positions) == 0:
    print("No missing values found.")
else:
    for row, col in missing_positions:
        print(
            f"Row {row}, "
            f"Column {col} ({df.columns[col]})"
        )


# ============================================================
# 4. Duplicate Analysis
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE ANALYSIS")
print("=" * 60)

duplicate_count = df.duplicated().sum()
duplicate_ratio = duplicate_count / len(df) * 100

print(f"Duplicate rows : {duplicate_count}")
print(f"Duplicate ratio: {duplicate_ratio:.2f}%")

if duplicate_count > 0:
    print("\nDuplicate rows:")
    display(
        df[df.duplicated(keep=False)]
        .sort_values(by=df.columns.tolist())
        .head(20)
    )


# ============================================================
# 5. Feature / Target Definition
# ============================================================

TARGET = "quality"

FEATURES = [
    col for col in df.columns
    if col != TARGET
]

X = df[FEATURES]
y = df[TARGET]

print("\n" + "=" * 60)
print("FEATURE / TARGET")
print("=" * 60)

print(f"Number of samples : {len(df)}")
print(f"Number of features: {len(FEATURES)}")
print(f"Target            : {TARGET}")

print("\nFeatures:")

for feature in FEATURES:
    print(f"- {feature}")


# ============================================================
# 6. Descriptive Statistics
# ============================================================

print("\n" + "=" * 60)
print("DESCRIPTIVE STATISTICS")
print("=" * 60)

display(
    df.describe().T
)


# ============================================================
# 7. Feature Range Analysis
# ============================================================

print("\n" + "=" * 60)
print("FEATURE RANGE ANALYSIS")
print("=" * 60)

range_table = pd.DataFrame({
    "Min": df.min(),
    "Max": df.max(),
    "Range": df.max() - df.min(),
    "Mean": df.mean(),
    "Std": df.std()
})

display(range_table)


# ============================================================
# 8. Unique Value Analysis
# ============================================================

print("\n" + "=" * 60)
print("UNIQUE VALUE ANALYSIS")
print("=" * 60)

unique_table = pd.DataFrame({
    "Unique Count": df.nunique(),
    "Unique Ratio (%)":
        df.nunique() / len(df) * 100
})

display(unique_table)


# ============================================================
# 9. Near-Constant Feature Analysis
# ============================================================

print("\n" + "=" * 60)
print("NEAR-CONSTANT FEATURE ANALYSIS")
print("=" * 60)

near_constant_features = []

for feature in FEATURES:

    value_ratio = (
        df[feature]
        .value_counts(normalize=True)
        .iloc[0]
    )

    if value_ratio >= 0.95:
        near_constant_features.append(
            (feature, value_ratio)
        )

if not near_constant_features:

    print("No near-constant features found.")

else:

    for feature, ratio in near_constant_features:

        print(
            f"{feature}: "
            f"{ratio * 100:.2f}% same value"
        )


# ============================================================
# 10. Target Distribution
# ============================================================

print("\n" + "=" * 60)
print("TARGET DISTRIBUTION")
print("=" * 60)

quality_counts = (
    df[TARGET]
    .value_counts()
    .sort_index()
)

quality_distribution = pd.DataFrame({
    "Count": quality_counts,
    "Ratio (%)":
        quality_counts / len(df) * 100
})

display(quality_distribution)


# ============================================================
# 11. Target Histogram
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    df[TARGET],
    bins=np.arange(
        df[TARGET].min() - 0.5,
        df[TARGET].max() + 1.5,
        1
    ),
    rwidth=0.8
)

plt.title("Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")

plt.xticks(
    sorted(df[TARGET].unique())
)

plt.tight_layout()
plt.show()


# ============================================================
# 12. Target Bar Chart
# ============================================================

plt.figure(figsize=(8, 5))

quality_counts.plot(
    kind="bar"
)

plt.title("Wine Quality Frequency")
plt.xlabel("Quality")
plt.ylabel("Number of Samples")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 13. Feature Histograms
# ============================================================

for feature in FEATURES:

    plt.figure(figsize=(7, 4))

    plt.hist(
        df[feature],
        bins=30
    )

    plt.title(
        f"Distribution of {feature}"
    )

    plt.xlabel(feature)
    plt.ylabel("Count")

    plt.tight_layout()
    plt.show()


# ============================================================
# 14. Feature Boxplots
# ============================================================

for feature in FEATURES:

    plt.figure(figsize=(7, 3))

    plt.boxplot(
        df[feature],
        vert=False
    )

    plt.title(
        f"Boxplot of {feature}"
    )

    plt.xlabel(feature)

    plt.tight_layout()
    plt.show()


# ============================================================
# 15. All Feature Boxplots
# ============================================================

plt.figure(figsize=(14, 7))

plt.boxplot(
    [df[feature] for feature in FEATURES],
    labels=FEATURES
)

plt.title(
    "Feature Distribution Comparison"
)

plt.ylabel("Value")

plt.xticks(rotation=60)

plt.tight_layout()
plt.show()


# ============================================================
# 16. Skewness Analysis
# ============================================================

print("\n" + "=" * 60)
print("SKEWNESS ANALYSIS")
print("=" * 60)

skewness = (
    df[FEATURES]
    .skew()
    .sort_values(
        ascending=False
    )
)

skew_table = pd.DataFrame({
    "Skewness": skewness
})

display(skew_table)


# ============================================================
# 17. Skewness Visualization
# ============================================================

plt.figure(figsize=(10, 5))

skewness.sort_values().plot(
    kind="barh"
)

plt.axvline(
    0,
    linewidth=1
)

plt.title(
    "Feature Skewness"
)

plt.xlabel("Skewness")

plt.tight_layout()
plt.show()


# ============================================================
# 18. Correlation Matrix
# ============================================================

print("\n" + "=" * 60)
print("CORRELATION MATRIX")
print("=" * 60)

corr = df.corr(
    numeric_only=True
)

display(corr)


# ============================================================
# 19. Correlation Heatmap
# ============================================================

plt.figure(figsize=(12, 9))

plt.imshow(
    corr,
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title(
    "Correlation Matrix"
)

plt.tight_layout()
plt.show()


# ============================================================
# 20. Feature ↔ Quality Correlation
# ============================================================

print("\n" + "=" * 60)
print("FEATURE ↔ QUALITY CORRELATION")
print("=" * 60)

quality_corr = (
    df[FEATURES + [TARGET]]
    .corr()[TARGET]
    .drop(TARGET)
    .sort_values()
)

display(
    quality_corr.to_frame(
        "Correlation"
    )
)


# ============================================================
# 21. Feature ↔ Quality Correlation Plot
# ============================================================

plt.figure(figsize=(8, 6))

quality_corr.plot(
    kind="barh"
)

plt.axvline(
    0,
    linewidth=1
)

plt.title(
    "Feature Correlation with Wine Quality"
)

plt.xlabel(
    "Correlation with Quality"
)

plt.tight_layout()
plt.show()


# ============================================================
# 22. Feature ↔ Quality Scatter Plot
# ============================================================

for feature in FEATURES:

    plt.figure(figsize=(6, 4))

    plt.scatter(
        df[feature],
        df[TARGET],
        alpha=0.3
    )

    plt.xlabel(feature)
    plt.ylabel(TARGET)

    plt.title(
        f"{feature} vs {TARGET}"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 23. Feature ↔ Quality + Linear Trend
# ============================================================

for feature in FEATURES:

    x = df[feature].values
    y = df[TARGET].values

    slope, intercept = np.polyfit(
        x,
        y,
        1
    )

    x_line = np.linspace(
        x.min(),
        x.max(),
        100
    )

    y_line = (
        slope * x_line +
        intercept
    )

    plt.figure(figsize=(6, 4))

    plt.scatter(
        x,
        y,
        alpha=0.25
    )

    plt.plot(
        x_line,
        y_line,
        linewidth=2
    )

    plt.xlabel(feature)
    plt.ylabel(TARGET)

    plt.title(
        f"{feature} vs {TARGET}"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 24. Feature Mean by Quality
# ============================================================

print("\n" + "=" * 60)
print("FEATURE MEAN BY QUALITY")
print("=" * 60)

quality_mean = (
    df.groupby(TARGET)[FEATURES]
      .mean()
)

display(
    quality_mean
)


# ============================================================
# 25. Feature Mean by Quality Visualization
# ============================================================

for feature in FEATURES:

    plt.figure(figsize=(7, 4))

    quality_mean[feature].plot(
        marker="o"
    )

    plt.title(
        f"Mean {feature} by Quality"
    )

    plt.xlabel("Quality")
    plt.ylabel(
        f"Mean {feature}"
    )

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 26. Feature Distribution by Quality
# ============================================================

for feature in FEATURES:

    plt.figure(figsize=(8, 5))

    df.boxplot(
        column=feature,
        by=TARGET
    )

    plt.title(
        f"{feature} Distribution by Quality"
    )

    plt.suptitle("")

    plt.xlabel("Quality")
    plt.ylabel(feature)

    plt.tight_layout()
    plt.show()


# ============================================================
# 27. Feature-to-Feature Correlation
# ============================================================

print("\n" + "=" * 60)
print("FEATURE-TO-FEATURE CORRELATION")
print("=" * 60)

feature_corr = (
    df[FEATURES]
    .corr()
    .abs()
)

upper_triangle = feature_corr.where(
    np.triu(
        np.ones(
            feature_corr.shape
        ),
        k=1
    ).astype(bool)
)

high_corr_pairs = (
    upper_triangle
    .stack()
    .sort_values(
        ascending=False
    )
)

display(
    high_corr_pairs
    .head(10)
    .to_frame(
        "Absolute Correlation"
    )
)


# ============================================================
# 28. Highly Correlated Feature Pairs
# ============================================================

top_pairs = high_corr_pairs.head(5)

for (feature1, feature2), corr_value in top_pairs.items():

    plt.figure(figsize=(6, 4))

    plt.scatter(
        df[feature1],
        df[feature2],
        alpha=0.3
    )

    plt.xlabel(feature1)
    plt.ylabel(feature2)

    plt.title(
        f"{feature1} vs {feature2}\n"
        f"|Correlation| = {corr_value:.3f}"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 29. IQR-based Outlier Detection
# ============================================================

print("\n" + "=" * 60)
print("IQR-BASED OUTLIER ANALYSIS")
print("=" * 60)

iqr_results = []

for feature in FEATURES:

    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (df[feature] < lower_bound) |
        (df[feature] > upper_bound)
    )

    outlier_count = (
        outlier_mask.sum()
    )

    outlier_ratio = (
        outlier_count /
        len(df) *
        100
    )

    iqr_results.append({

        "Feature": feature,

        "Q1": Q1,

        "Q3": Q3,

        "IQR": IQR,

        "Lower Bound":
            lower_bound,

        "Upper Bound":
            upper_bound,

        "Outlier Count":
            outlier_count,

        "Outlier Ratio (%)":
            outlier_ratio
    })


iqr_table = pd.DataFrame(
    iqr_results
)

display(
    iqr_table
)


# ============================================================
# 30. Number of Outlier Features per Sample
# ============================================================

outlier_count_per_row = pd.Series(
    0,
    index=df.index
)

for feature in FEATURES:

    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    mask = (
        (df[feature] < lower) |
        (df[feature] > upper)
    )

    outlier_count_per_row += (
        mask.astype(int)
    )


print("\n" + "=" * 60)
print("OUTLIER FEATURES PER SAMPLE")
print("=" * 60)

outlier_count_distribution = (
    outlier_count_per_row
    .value_counts()
    .sort_index()
    .to_frame(
        "Sample Count"
    )
)

display(
    outlier_count_distribution
)


# ============================================================
# 31. Samples Containing IQR Outliers
# ============================================================

all_iqr_outlier_mask = pd.Series(
    False,
    index=df.index
)

for feature in FEATURES:

    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    mask = (
        (df[feature] < lower) |
        (df[feature] > upper)
    )

    all_iqr_outlier_mask |= mask


print(
    "Samples containing at least "
    f"one IQR outlier: "
    f"{all_iqr_outlier_mask.sum()}"
)


# ============================================================
# 32. IQR Outlier Samples
# ============================================================

iqr_outlier_rows = df[
    all_iqr_outlier_mask
]

print("\n===== IQR Outlier Samples =====")

display(
    iqr_outlier_rows
    .head(20)
)


# ============================================================
# 33. EDA Summary
# ============================================================

print("\n" + "=" * 60)
print("EDA SUMMARY")
print("=" * 60)

print("\n[Dataset]")
print(
    f"- Samples : {len(df):,}"
)
print(
    f"- Features: {len(FEATURES)}"
)
print(
    f"- Target  : {TARGET}"
)

print("\n[Missing Values]")
print(
    f"- Total missing values: "
    f"{df.isnull().sum().sum()}"
)

print("\n[Duplicates]")
print(
    f"- Duplicate rows: "
    f"{df.duplicated().sum()}"
)

print("\n[Target]")
print(
    f"- Target range: "
    f"{df[TARGET].min()} ~ "
    f"{df[TARGET].max()}"
)

print(
    f"- Target mean: "
    f"{df[TARGET].mean():.4f}"
)

print(
    f"- Target std: "
    f"{df[TARGET].std():.4f}"
)

print("\n[Skewness]")

print(
    f"- Most positively skewed: "
    f"{skewness.idxmax()} "
    f"({skewness.max():.3f})"
)

print(
    f"- Most negatively skewed: "
    f"{skewness.idxmin()} "
    f"({skewness.min():.3f})"
)

print("\n[Feature ↔ Quality Correlation]")

highest_positive = (
    quality_corr.idxmax()
)

highest_negative = (
    quality_corr.idxmin()
)

print(
    f"- Highest positive: "
    f"{highest_positive} "
    f"({quality_corr.max():.3f})"
)

print(
    f"- Highest negative: "
    f"{highest_negative} "
    f"({quality_corr.min():.3f})"
)

print("\n[IQR Outlier]")

print(
    "- Samples containing at least "
    f"one IQR outlier: "
    f"{all_iqr_outlier_mask.sum()}"
)

print("\n" + "=" * 60)
print("EDA COMPLETE")
print("=" * 60)
```
