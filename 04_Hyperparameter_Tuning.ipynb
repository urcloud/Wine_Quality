```python
# ============================================================
# WINE QUALITY PREDICTION - MODEL TUNING
# ============================================================


# ============================================================
# 0. IMPORT
# ============================================================

import os
import random
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import KFold, cross_validate, GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from xgboost import XGBRegressor

import tensorflow as tf

warnings.filterwarnings("ignore")


print("=" * 70)
print("WINE QUALITY PREDICTION - MODEL TUNING")
print("=" * 70)


# ============================================================
# 1. RANDOM SEED
# ============================================================

RANDOM_STATE = 42

os.environ["PYTHONHASHSEED"] = str(RANDOM_STATE)

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
tf.keras.utils.set_random_seed(RANDOM_STATE)

print(f"Random State : {RANDOM_STATE}")


# ============================================================
# 2. LOAD PREPROCESSING RESULTS
# ============================================================

print("\n" + "=" * 70)
print("2. LOAD PREPROCESSING RESULTS")
print("=" * 70)


PREPROCESSING_PATH = "preprocessed_data/wine_preprocessed.pkl"


if not os.path.exists(PREPROCESSING_PATH):
    raise FileNotFoundError(
        f"""
Preprocessing file not found:

{PREPROCESSING_PATH}

Please run 02_Data_Processing.ipynb first.
"""
    )


data = joblib.load(PREPROCESSING_PATH)


# ------------------------------------------------------------
# Restore variables
# ------------------------------------------------------------

RANDOM_STATE = data["RANDOM_STATE"]

FEATURES = data["FEATURES"]
TARGET = data["TARGET"]

X_train = data["X_train"]
y_train = data["y_train"]

X_val = data["X_val"]
y_val = data["y_val"]

X_test = data["X_test"]
y_test = data["y_test"]


print("Loaded preprocessing results from:")
print(f"  {PREPROCESSING_PATH}")

print(f"\nTrain      : {X_train.shape}")
print(f"Validation : {X_val.shape}")
print(f"Test       : {X_test.shape}")
print(f"Features   : {len(FEATURES)}")


# ============================================================
# 3. COMBINE TRAIN + VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("3. PREPARE TUNING DATA")
print("=" * 70)


# ------------------------------------------------------------
# Hyperparameter tuning is performed using
# Train + Validation only.
#
# Test set remains untouched until final evaluation.
# ------------------------------------------------------------

X_train_val = pd.concat(
    [X_train, X_val],
    axis=0
).reset_index(drop=True)


y_train_val = pd.concat(
    [y_train, y_val],
    axis=0
).reset_index(drop=True)


print(f"Train + Validation : {X_train_val.shape}")
print(f"Test               : {X_test.shape}")


# ============================================================
# 4. CROSS VALIDATION SETUP
# ============================================================

print("\n" + "=" * 70)
print("4. CROSS VALIDATION")
print("=" * 70)


CV = KFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)


print("CV Strategy : 5-Fold K-Fold")
print("Shuffle     : True")
print(f"Random State: {RANDOM_STATE}")


# ============================================================
# 5. BASELINE MODELS
# ============================================================

print("\n" + "=" * 70)
print("5. BASELINE MODELS")
print("=" * 70)


baseline_models = {

    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        max_features=1.0,
        min_samples_leaf=1,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.03,
        max_depth=2,
        min_samples_leaf=2,
        random_state=RANDOM_STATE
    ),

    "XGBoost": XGBRegressor(
        n_estimators=300,
        learning_rate=0.03,
        max_depth=3,
        min_child_weight=1,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.0,
        reg_lambda=1.0,
        objective="reg:squarederror",
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}


print("Models:")
for name in baseline_models:
    print(f"- {name}")


# ============================================================
# 6. CROSS VALIDATION BASELINE
# ============================================================

print("\n" + "=" * 70)
print("6. CROSS VALIDATION BASELINE")
print("=" * 70)


scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}


baseline_cv_results = []


for model_name, model in baseline_models.items():

    print(f"\n[{model_name}]")

    cv_result = cross_validate(
        model,
        X_train_val,
        y_train_val,
        cv=CV,
        scoring=scoring,
        n_jobs=-1,
        return_train_score=False
    )

    mae = -cv_result["test_MAE"].mean()
    rmse = -cv_result["test_RMSE"].mean()
    r2 = cv_result["test_R2"].mean()

    baseline_cv_results.append({
        "Model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

    print(f"CV MAE  : {mae:.4f}")
    print(f"CV RMSE : {rmse:.4f}")
    print(f"CV R²   : {r2:.4f}")


baseline_cv_df = pd.DataFrame(baseline_cv_results)

baseline_cv_df = baseline_cv_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)


print("\n" + "-" * 70)
print("BASELINE CROSS VALIDATION RESULTS")
print("-" * 70)

display(
    baseline_cv_df.round(4)
)


# ============================================================
# 7. RANDOM FOREST GRID SEARCH
# ============================================================

print("\n" + "=" * 70)
print("7. RANDOM FOREST HYPERPARAMETER TUNING")
print("=" * 70)


rf = RandomForestRegressor(
    random_state=RANDOM_STATE,
    n_jobs=-1
)


rf_param_grid = {

    "n_estimators": [
        200,
        300,
        500
    ],

    "max_depth": [
        None,
        5,
        10,
        15
    ],

    "min_samples_leaf": [
        1,
        2,
        4
    ],

    "max_features": [
        0.7,
        1.0
    ]
}


rf_grid = GridSearchCV(
    estimator=rf,
    param_grid=rf_param_grid,
    scoring="neg_root_mean_squared_error",
    cv=CV,
    n_jobs=-1,
    refit=True,
    verbose=1,
    return_train_score=False
)


rf_grid.fit(
    X_train_val,
    y_train_val
)


print("\nRandom Forest tuning complete.")

print("\nBest Parameters:")
print(rf_grid.best_params_)

print(f"\nBest CV RMSE: {-rf_grid.best_score_:.4f}")


# ============================================================
# 8. GRADIENT BOOSTING GRID SEARCH
# ============================================================

print("\n" + "=" * 70)
print("8. GRADIENT BOOSTING HYPERPARAMETER TUNING")
print("=" * 70)


gb = GradientBoostingRegressor(
    random_state=RANDOM_STATE
)


gb_param_grid = {

    "n_estimators": [
        100,
        200,
        300
    ],

    "learning_rate": [
        0.01,
        0.03,
        0.05,
        0.1
    ],

    "max_depth": [
        1,
        2,
        3
    ],

    "min_samples_leaf": [
        1,
        2,
        4
    ]
}


gb_grid = GridSearchCV(
    estimator=gb,
    param_grid=gb_param_grid,
    scoring="neg_root_mean_squared_error",
    cv=CV,
    n_jobs=-1,
    refit=True,
    verbose=1,
    return_train_score=False
)


gb_grid.fit(
    X_train_val,
    y_train_val
)


print("\nGradient Boosting tuning complete.")

print("\nBest Parameters:")
print(gb_grid.best_params_)

print(f"\nBest CV RMSE: {-gb_grid.best_score_:.4f}")


# ============================================================
# 9. XGBOOST GRID SEARCH
# ============================================================

print("\n" + "=" * 70)
print("9. XGBOOST HYPERPARAMETER TUNING")
print("=" * 70)


xgb = XGBRegressor(
    objective="reg:squarederror",
    random_state=RANDOM_STATE,
    n_jobs=-1
)


xgb_param_grid = {

    "n_estimators": [
        200,
        300,
        500
    ],

    "learning_rate": [
        0.01,
        0.03,
        0.05
    ],

    "max_depth": [
        2,
        3,
        4
    ],

    "min_child_weight": [
        1,
        3,
        5
    ],

    "subsample": [
        0.8,
        1.0
    ],

    "colsample_bytree": [
        0.8,
        1.0
    ]
}


xgb_grid = GridSearchCV(
    estimator=xgb,
    param_grid=xgb_param_grid,
    scoring="neg_root_mean_squared_error",
    cv=CV,
    n_jobs=-1,
    refit=True,
    verbose=1,
    return_train_score=False
)


xgb_grid.fit(
    X_train_val,
    y_train_val
)


print("\nXGBoost tuning complete.")

print("\nBest Parameters:")
print(xgb_grid.best_params_)

print(f"\nBest CV RMSE: {-xgb_grid.best_score_:.4f}")


# ============================================================
# 10. TUNING RESULT SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("10. TUNING RESULT SUMMARY")
print("=" * 70)


tuning_results = []


# ------------------------------------------------------------
# Random Forest
# ------------------------------------------------------------

tuning_results.append({
    "Model": "Random Forest",
    "CV RMSE": -rf_grid.best_score_,
    "Best Parameters": str(rf_grid.best_params_)
})


# ------------------------------------------------------------
# Gradient Boosting
# ------------------------------------------------------------

tuning_results.append({
    "Model": "Gradient Boosting",
    "CV RMSE": -gb_grid.best_score_,
    "Best Parameters": str(gb_grid.best_params_)
})


# ------------------------------------------------------------
# XGBoost
# ------------------------------------------------------------

tuning_results.append({
    "Model": "XGBoost",
    "CV RMSE": -xgb_grid.best_score_,
    "Best Parameters": str(xgb_grid.best_params_)
})


tuning_summary = pd.DataFrame(
    tuning_results
).sort_values(
    by="CV RMSE",
    ascending=True
).reset_index(drop=True)


display(
    tuning_summary.round(4)
)


# ============================================================
# 11. SELECT TUNED MODEL
# ============================================================

print("\n" + "=" * 70)
print("11. SELECT TUNED MODEL")
print("=" * 70)


best_model_name = tuning_summary.iloc[0]["Model"]


if best_model_name == "Random Forest":

    best_tuned_model = rf_grid.best_estimator_

    best_params = rf_grid.best_params_


elif best_model_name == "Gradient Boosting":

    best_tuned_model = gb_grid.best_estimator_

    best_params = gb_grid.best_params_


elif best_model_name == "XGBoost":

    best_tuned_model = xgb_grid.best_estimator_

    best_params = xgb_grid.best_params_


else:

    raise ValueError(
        f"Unknown model: {best_model_name}"
    )


print(f"Selected Model : {best_model_name}")

print("\nSelected Parameters:")

for key, value in best_params.items():
    print(f"{key}: {value}")


# ============================================================
# 12. FIT FINAL TUNED MODEL
# ============================================================

print("\n" + "=" * 70)
print("12. FIT FINAL TUNED MODEL")
print("=" * 70)


# GridSearchCV(refit=True) already fitted the model
# on the complete Train + Validation dataset.

final_tuned_model = best_tuned_model


print("Final tuned model is ready.")
print(f"Training data : {X_train_val.shape}")


# ============================================================
# 13. FINAL TEST PREDICTION
# ============================================================

print("\n" + "=" * 70)
print("13. FINAL TEST EVALUATION")
print("=" * 70)


y_test_pred = final_tuned_model.predict(
    X_test
)


test_mae = mean_absolute_error(
    y_test,
    y_test_pred
)


test_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred
    )
)


test_r2 = r2_score(
    y_test,
    y_test_pred
)


print(f"Model : {best_model_name}")

print(f"\nTest MAE  : {test_mae:.4f}")
print(f"Test RMSE : {test_rmse:.4f}")
print(f"Test R²   : {test_r2:.4f}")


# ============================================================
# 14. LOAD 03 BASELINE RESULT
# ============================================================

print("\n" + "=" * 70)
print("14. LOAD BASELINE RESULT")
print("=" * 70)


BASELINE_TEST_PATH = "wine_final_test_results.csv"


if not os.path.exists(BASELINE_TEST_PATH):

    raise FileNotFoundError(
        f"""
Baseline result file not found:

{BASELINE_TEST_PATH}

Please run 03_Modeling.ipynb first.
"""
    )


baseline_test_df = pd.read_csv(
    BASELINE_TEST_PATH
)


print("Baseline result loaded.")

print("\nColumns:")
print(baseline_test_df.columns.tolist())

print("\nBaseline result:")
display(baseline_test_df)


# ------------------------------------------------------------
# Expected structure
#
# Metric | Validation | Test
# MAE    | ...        | ...
# RMSE   | ...        | ...
# R²     | ...        | ...
# ------------------------------------------------------------

required_columns = {
    "Metric",
    "Validation",
    "Test"
}


if not required_columns.issubset(
    baseline_test_df.columns
):

    raise ValueError(
        f"""
Unexpected baseline result format.

Expected columns:
{sorted(required_columns)}

Actual columns:
{baseline_test_df.columns.tolist()}
"""
    )


# ------------------------------------------------------------
# Extract baseline metrics
# ------------------------------------------------------------

def get_baseline_metric(metric_name, split_name):

    row = baseline_test_df[
        baseline_test_df["Metric"] == metric_name
    ]

    if len(row) == 0:

        raise ValueError(
            f"Metric '{metric_name}' was not found "
            "in the baseline result."
        )

    return float(
        row.iloc[0][split_name]
    )


baseline_val_mae = get_baseline_metric(
    "MAE",
    "Validation"
)

baseline_val_rmse = get_baseline_metric(
    "RMSE",
    "Validation"
)

baseline_val_r2 = get_baseline_metric(
    "R²",
    "Validation"
)


baseline_test_mae = get_baseline_metric(
    "MAE",
    "Test"
)

baseline_test_rmse = get_baseline_metric(
    "RMSE",
    "Test"
)

baseline_test_r2 = get_baseline_metric(
    "R²",
    "Test"
)


print("\nBaseline Validation:")
print(f"MAE  : {baseline_val_mae:.4f}")
print(f"RMSE : {baseline_val_rmse:.4f}")
print(f"R²   : {baseline_val_r2:.4f}")


print("\nBaseline Test:")
print(f"MAE  : {baseline_test_mae:.4f}")
print(f"RMSE : {baseline_test_rmse:.4f}")
print(f"R²   : {baseline_test_r2:.4f}")


# ============================================================
# 15. BASELINE vs TUNED CV
# ============================================================

print("\n" + "=" * 70)
print("15. BASELINE vs TUNED CV")
print("=" * 70)


# ------------------------------------------------------------
# IMPORTANT
#
# Baseline Validation RMSE from 03_Modeling:
#     0.635695
#
# Tuned model:
#     5-Fold CV RMSE
#
# These are not exactly the same evaluation protocol.
#
# Therefore, this comparison is used as a reference,
# not as a direct final performance comparison.
# ------------------------------------------------------------

tuned_cv_rmse = tuning_summary.iloc[0]["CV RMSE"]


cv_comparison_df = pd.DataFrame({

    "Model": [
        "Baseline Random Forest",
        f"Tuned {best_model_name}"
    ],

    "RMSE": [
        baseline_val_rmse,
        tuned_cv_rmse
    ]

})


display(
    cv_comparison_df.round(4)
)


print(f"\nBaseline Validation RMSE : {baseline_val_rmse:.4f}")
print(f"Tuned 5-Fold CV RMSE     : {tuned_cv_rmse:.4f}")


cv_rmse_change = (
    tuned_cv_rmse
    - baseline_val_rmse
)


print(
    f"RMSE Difference           : "
    f"{cv_rmse_change:+.4f}"
)


# ============================================================
# 16. BASELINE vs TUNED TEST
# ============================================================

print("\n" + "=" * 70)
print("16. BASELINE vs TUNED TEST")
print("=" * 70)


comparison_df = pd.DataFrame({

    "Metric": [
        "MAE",
        "RMSE",
        "R²"
    ],

    "Baseline Random Forest": [
        baseline_test_mae,
        baseline_test_rmse,
        baseline_test_r2
    ],

    f"Tuned {best_model_name}": [
        test_mae,
        test_rmse,
        test_r2
    ]

})


display(
    comparison_df.round(4)
)


# ============================================================
# 17. TEST PERFORMANCE CHANGE
# ============================================================

print("\n" + "=" * 70)
print("17. TEST PERFORMANCE CHANGE")
print("=" * 70)


mae_change = (
    test_mae
    - baseline_test_mae
)


rmse_change = (
    test_rmse
    - baseline_test_rmse
)


r2_change = (
    test_r2
    - baseline_test_r2
)


print(f"MAE  Change : {mae_change:+.4f}")
print(f"RMSE Change : {rmse_change:+.4f}")
print(f"R²   Change : {r2_change:+.4f}")


print("\nRelative Change:")


print(
    f"MAE  : "
    f"{(mae_change / baseline_test_mae) * 100:+.2f}%"
)


print(
    f"RMSE : "
    f"{(rmse_change / baseline_test_rmse) * 100:+.2f}%"
)


print(
    f"R²   : "
    f"{(r2_change / abs(baseline_test_r2)) * 100:+.2f}%"
)


# ============================================================
# 18. ACTUAL vs PREDICTED
# ============================================================

print("\n" + "=" * 70)
print("18. ACTUAL vs PREDICTED")
print("=" * 70)


plt.figure(figsize=(7, 7))


plt.scatter(
    y_test,
    y_test_pred,
    alpha=0.7
)


min_value = min(
    y_test.min(),
    y_test_pred.min()
)


max_value = max(
    y_test.max(),
    y_test_pred.max()
)


plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)


plt.xlabel("Actual Quality")
plt.ylabel("Predicted Quality")

plt.title(
    f"Actual vs Predicted - Tuned {best_model_name}"
)


plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 19. RESIDUAL ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("19. RESIDUAL ANALYSIS")
print("=" * 70)


residuals = y_test - y_test_pred


residual_df = pd.DataFrame({

    "Actual": y_test.values,

    "Predicted": y_test_pred,

    "Residual": residuals.values,

    "Absolute Error":
        np.abs(residuals.values)

})


largest_errors = residual_df.sort_values(
    by="Absolute Error",
    ascending=False
).head(20)


print("Largest prediction errors:")


display(
    largest_errors.round(4)
)


# ============================================================
# 20. RESIDUAL DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("20. RESIDUAL DISTRIBUTION")
print("=" * 70)


plt.figure(figsize=(8, 5))


plt.hist(
    residuals,
    bins=20,
    edgecolor="black"
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel("Residual")
plt.ylabel("Frequency")


plt.title(
    f"Residual Distribution - Tuned {best_model_name}"
)


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()

plt.show()


# ============================================================
# 21. RESIDUAL vs PREDICTED
# ============================================================

print("\n" + "=" * 70)
print("21. RESIDUAL vs PREDICTED")
print("=" * 70)


plt.figure(figsize=(8, 5))


plt.scatter(
    y_test_pred,
    residuals,
    alpha=0.7
)


plt.axhline(
    0,
    linestyle="--"
)


plt.xlabel("Predicted Quality")
plt.ylabel("Residual")


plt.title(
    f"Residual vs Predicted - Tuned {best_model_name}"
)


plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 22. FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 70)
print("22. FEATURE IMPORTANCE")
print("=" * 70)


if hasattr(
    final_tuned_model,
    "feature_importances_"
):

    feature_importance_df = pd.DataFrame({

        "Feature": FEATURES,

        "Importance":
            final_tuned_model.feature_importances_

    })


    feature_importance_df = (
        feature_importance_df
        .sort_values(
            by="Importance",
            ascending=False
        )
        .reset_index(drop=True)
    )


    display(
        feature_importance_df.round(4)
    )


    plt.figure(
        figsize=(9, 6)
    )


    plt.barh(
        feature_importance_df["Feature"],
        feature_importance_df["Importance"]
    )


    plt.gca().invert_yaxis()


    plt.xlabel("Importance")

    plt.ylabel("Feature")

    plt.title(
        f"Feature Importance - Tuned {best_model_name}"
    )


    plt.tight_layout()

    plt.show()


else:

    feature_importance_df = None

    print(
        "Feature importance is not available for this model."
    )


# ============================================================
# 23. PREDICTION ERROR BY ACTUAL QUALITY
# ============================================================

print("\n" + "=" * 70)
print("23. ERROR BY ACTUAL QUALITY")
print("=" * 70)


error_by_quality = residual_df.copy()


error_by_quality["Quality"] = (
    error_by_quality["Actual"]
    .astype(int)
)


error_summary = (
    error_by_quality
    .groupby("Quality")
    .agg(
        Count=("Absolute Error", "count"),

        MAE=("Absolute Error", "mean"),

        Mean_Residual=("Residual", "mean"),

        RMSE=(
            "Residual",
            lambda x: np.sqrt(
                np.mean(x ** 2)
            )
        )
    )
    .reset_index()
)


display(
    error_summary.round(4)
)


# ============================================================
# 24. SAVE TUNING RESULTS
# ============================================================

print("\n" + "=" * 70)
print("24. SAVE RESULTS")
print("=" * 70)


tuning_output_dir = "tuning_results"


os.makedirs(
    tuning_output_dir,
    exist_ok=True
)


# ------------------------------------------------------------
# CV baseline
# ------------------------------------------------------------

baseline_cv_df.to_csv(
    os.path.join(
        tuning_output_dir,
        "baseline_cv_results.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Tuning summary
# ------------------------------------------------------------

tuning_summary.to_csv(
    os.path.join(
        tuning_output_dir,
        "tuning_summary.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Baseline vs tuned test comparison
# ------------------------------------------------------------

comparison_df.to_csv(
    os.path.join(
        tuning_output_dir,
        "baseline_vs_tuned.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Test predictions
# ------------------------------------------------------------

test_prediction_df = pd.DataFrame({

    "Actual": y_test.values,

    "Predicted": y_test_pred,

    "Residual": residuals.values,

    "Absolute_Error":
        np.abs(residuals.values)

})


test_prediction_df.to_csv(
    os.path.join(
        tuning_output_dir,
        "tuned_test_predictions.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Error by quality
# ------------------------------------------------------------

error_summary.to_csv(
    os.path.join(
        tuning_output_dir,
        "error_by_quality.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Feature importance
# ------------------------------------------------------------

if feature_importance_df is not None:

    feature_importance_df.to_csv(
        os.path.join(
            tuning_output_dir,
            "tuned_feature_importance.csv"
        ),
        index=False
    )


# ------------------------------------------------------------
# Best parameters
# ------------------------------------------------------------

best_params_df = pd.DataFrame(
    [
        {
            "Model": best_model_name,
            **best_params
        }
    ]
)


best_params_df.to_csv(
    os.path.join(
        tuning_output_dir,
        "best_parameters.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Baseline metric summary
# ------------------------------------------------------------

baseline_summary_df = pd.DataFrame({

    "Metric": [
        "MAE",
        "RMSE",
        "R²"
    ],

    "Validation": [
        baseline_val_mae,
        baseline_val_rmse,
        baseline_val_r2
    ],

    "Test": [
        baseline_test_mae,
        baseline_test_rmse,
        baseline_test_r2
    ]

})


baseline_summary_df.to_csv(
    os.path.join(
        tuning_output_dir,
        "baseline_metrics.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Final tuned test metrics
# ------------------------------------------------------------

tuned_test_metrics_df = pd.DataFrame({

    "Metric": [
        "MAE",
        "RMSE",
        "R²"
    ],

    "Test": [
        test_mae,
        test_rmse,
        test_r2
    ]

})


tuned_test_metrics_df.to_csv(
    os.path.join(
        tuning_output_dir,
        "tuned_test_metrics.csv"
    ),
    index=False
)


print(
    f"Results saved to: {tuning_output_dir}/"
)


# ============================================================
# 25. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("TUNING COMPLETE")
print("=" * 70)


print(f"Selected Model : {best_model_name}")


print("\nBest Parameters:")


for key, value in best_params.items():

    print(
        f"  {key}: {value}"
    )


print("\nFinal Tuned Test Performance:")


print(
    f"  MAE  : {test_mae:.4f}"
)


print(
    f"  RMSE : {test_rmse:.4f}"
)


print(
    f"  R²   : {test_r2:.4f}"
)


print("\nBaseline Random Forest Test Performance:")


print(
    f"  MAE  : {baseline_test_mae:.4f}"
)


print(
    f"  RMSE : {baseline_test_rmse:.4f}"
)


print(
    f"  R²   : {baseline_test_r2:.4f}"
)


print("\nEvaluation Protocol:")


print(
    "  - Hyperparameter tuning: Train + Validation"
)


print(
    "  - Cross Validation: 5-Fold"
)


print(
    "  - Test set: Final evaluation only"
)


print("=" * 70)
```
