```python
# ============================================================
# MODELING
#
# Models
#   1. Linear Regression
#   2. Ridge
#   3. Decision Tree
#   4. Random Forest
#   5. Gradient Boosting
#   6. XGBoost
#   7. DNN
#
# Dataset Comparison
#   1. Original
#   2. IQR Removed
#   3. Autoencoder Removed
#
# Evaluation
#   - MAE
#   - RMSE
#   - R²
#
# Validation → Model Selection
# Test       → Final Evaluation
# ============================================================


# ============================================================
# 0. IMPORT
# ============================================================

import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone

# ------------------------------------------------------------
# Linear Models
# ------------------------------------------------------------

from sklearn.linear_model import (
    LinearRegression,
    Ridge
)

# ------------------------------------------------------------
# Tree / Ensemble Models
# ------------------------------------------------------------

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

# ------------------------------------------------------------
# Evaluation Metrics
# ------------------------------------------------------------

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# XGBoost
# ------------------------------------------------------------

from xgboost import XGBRegressor

# ------------------------------------------------------------
# TensorFlow / Keras
# ------------------------------------------------------------

import tensorflow as tf


# ============================================================
# 1. LOAD PREPROCESSING RESULTS
# ============================================================

print("=" * 70)
print("WINE QUALITY PREDICTION - MODELING")
print("=" * 70)


# ------------------------------------------------------------
# Random Seed
# ------------------------------------------------------------

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


# ------------------------------------------------------------
# Preprocessing Result Path
# ------------------------------------------------------------

PREPROCESSING_PATH = os.path.join(
    "preprocessed_data",
    "wine_preprocessed.pkl"
)


print("\n" + "=" * 70)
print("1. LOAD PREPROCESSING RESULTS")
print("=" * 70)


if not os.path.exists(PREPROCESSING_PATH):

    raise FileNotFoundError(
        f"Preprocessed data file not found:\n"
        f"{PREPROCESSING_PATH}\n\n"
        f"Please run 02_Data_Preprocessing.ipynb first."
    )


# ------------------------------------------------------------
# Load
# ------------------------------------------------------------

preprocessed_data = joblib.load(
    PREPROCESSING_PATH
)


print(
    f"Loaded preprocessing results from:"
)

print(
    f"  {PREPROCESSING_PATH}"
)


# ============================================================
# 2. RESTORE PREPROCESSING VARIABLES
# ============================================================
#
# 02_Data_Preprocessing.ipynb에서 저장한 결과를
# 03_Modeling.ipynb에서 사용할 변수명으로 복원한다.
#
# ============================================================

print("\n" + "=" * 70)
print("2. RESTORE PREPROCESSING VARIABLES")
print("=" * 70)


# ------------------------------------------------------------
# Basic Information
# ------------------------------------------------------------

RANDOM_STATE = preprocessed_data["RANDOM_STATE"]

FEATURES = preprocessed_data["FEATURES"]

TARGET = preprocessed_data["TARGET"]


# ------------------------------------------------------------
# Cleaned Dataset
# ------------------------------------------------------------

df_clean = preprocessed_data["df_clean"]


# ------------------------------------------------------------
# Train / Validation / Test
# ------------------------------------------------------------

X_train = preprocessed_data["X_train"]
y_train = preprocessed_data["y_train"]

X_val = preprocessed_data["X_val"]
y_val = preprocessed_data["y_val"]

X_test = preprocessed_data["X_test"]
y_test = preprocessed_data["y_test"]


# ------------------------------------------------------------
# StandardScaler Results
# ------------------------------------------------------------

X_train_scaled = preprocessed_data[
    "X_train_scaled"
]

X_val_scaled = preprocessed_data[
    "X_val_scaled"
]

X_test_scaled = preprocessed_data[
    "X_test_scaled"
]


standard_scaler = preprocessed_data[
    "standard_scaler"
]


# ------------------------------------------------------------
# MinMaxScaler Results
# ------------------------------------------------------------

X_train_minmax = preprocessed_data[
    "X_train_minmax"
]

X_val_minmax = preprocessed_data[
    "X_val_minmax"
]

X_test_minmax = preprocessed_data[
    "X_test_minmax"
]


minmax_scaler = preprocessed_data[
    "minmax_scaler"
]


# ------------------------------------------------------------
# IQR Outlier
# ------------------------------------------------------------

iqr_outlier_mask_train = preprocessed_data[
    "iqr_outlier_mask_train"
]

iqr_outlier_count_train = preprocessed_data[
    "iqr_outlier_count_train"
]

iqr_table = preprocessed_data[
    "iqr_table"
]


# ------------------------------------------------------------
# Autoencoder Outlier
# ------------------------------------------------------------

ae_outlier_mask_train = preprocessed_data[
    "ae_outlier_mask_train"
]

ae_outlier_mask_val = preprocessed_data[
    "ae_outlier_mask_val"
]

ae_outlier_mask_test = preprocessed_data[
    "ae_outlier_mask_test"
]


# ------------------------------------------------------------
# Autoencoder Reconstruction Error
# ------------------------------------------------------------

train_reconstruction_error = preprocessed_data[
    "train_reconstruction_error"
]

val_reconstruction_error = preprocessed_data[
    "val_reconstruction_error"
]

test_reconstruction_error = preprocessed_data[
    "test_reconstruction_error"
]


# ------------------------------------------------------------
# Autoencoder Threshold
# ------------------------------------------------------------

ae_threshold = preprocessed_data[
    "ae_threshold"
]


# ------------------------------------------------------------
# Modeling Datasets
# ------------------------------------------------------------

X_train_original = preprocessed_data[
    "X_train_original"
]

y_train_original = preprocessed_data[
    "y_train_original"
]

X_train_iqr_clean = preprocessed_data[
    "X_train_iqr_clean"
]

y_train_iqr_clean = preprocessed_data[
    "y_train_iqr_clean"
]

X_train_ae_clean = preprocessed_data[
    "X_train_ae_clean"
]

y_train_ae_clean = preprocessed_data[
    "y_train_ae_clean"
]


print("Preprocessing variables restored successfully.")


# ============================================================
# 3. BASIC DATA INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("3. DATA INFORMATION")
print("=" * 70)


print(f"Train      : {X_train.shape}")
print(f"Validation : {X_val.shape}")
print(f"Test       : {X_test.shape}")

print(f"Features   : {X_train.shape[1]}")

print(f"Target     : {TARGET}")


# ============================================================
# 4. MODELING DATASET
# ============================================================
#
# IMPORTANT
# ------------------------------------------------------------
# Validation / Test data are NOT removed.
#
# Training datasets were already created in
# 02_Data_Preprocessing.ipynb.
#
# Original       : 951
# IQR Removed    : 719
# AE Removed     : 935
#
# Validation     : 204
# Test           : 204
#
# ============================================================

print("\n" + "=" * 70)
print("4. MODELING DATASET")
print("=" * 70)


# ------------------------------------------------------------
# Dataset Dictionary
# ------------------------------------------------------------

train_datasets = {

    "Original": (
        X_train_original,
        y_train_original
    ),

    "IQR Removed": (
        X_train_iqr_clean,
        y_train_iqr_clean
    ),

    "AE Removed": (
        X_train_ae_clean,
        y_train_ae_clean
    )
}


# ------------------------------------------------------------
# Dataset Information
# ------------------------------------------------------------

dataset_info = []


for dataset_name, (X_data, y_data) in train_datasets.items():

    dataset_info.append({

        "Dataset":
            dataset_name,

        "Samples":
            len(X_data),

        "Removed":
            len(X_train) - len(X_data),

        "Remaining Ratio (%)":
            len(X_data) / len(X_train) * 100

    })


dataset_info_df = pd.DataFrame(
    dataset_info
)


display(
    dataset_info_df.round(2)
)


# ============================================================
# 5. STANDARD SCALED DATASET
# ============================================================
#
# Linear Regression / Ridge / DNN
#     → StandardScaler
#
# Tree / Ensemble Models
#     → Raw Features
#
# StandardScaler 결과는
# 02_Data_Preprocessing.ipynb에서 이미 계산되었다.
#
# ============================================================

print("\n" + "=" * 70)
print("5. SCALED DATASET")
print("=" * 70)


# ------------------------------------------------------------
# Saved StandardScaler Results
# ------------------------------------------------------------

X_train_scaled_df = X_train_scaled.copy()

X_val_scaled_df = X_val_scaled.copy()

X_test_scaled_df = X_test_scaled.copy()


# ------------------------------------------------------------
# Original
# ------------------------------------------------------------

X_train_original_scaled = (
    X_train_scaled_df.copy()
)


# ------------------------------------------------------------
# IQR Removed
# ------------------------------------------------------------

X_train_iqr_scaled = (
    X_train_scaled_df.loc[
        ~iqr_outlier_mask_train
    ].copy()
)


# ------------------------------------------------------------
# Autoencoder Removed
# ------------------------------------------------------------

X_train_ae_scaled = (
    X_train_scaled_df.loc[
        ~ae_outlier_mask_train
    ].copy()
)


# ------------------------------------------------------------
# Scaled Dataset Dictionary
# ------------------------------------------------------------

scaled_train_datasets = {

    "Original": (
        X_train_original_scaled,
        y_train_original
    ),

    "IQR Removed": (
        X_train_iqr_scaled,
        y_train_iqr_clean
    ),

    "AE Removed": (
        X_train_ae_scaled,
        y_train_ae_clean
    )
}


print(
    "Train scaled shape:",
    X_train_scaled_df.shape
)

print(
    "Validation scaled shape:",
    X_val_scaled_df.shape
)

print(
    "Test scaled shape:",
    X_test_scaled_df.shape
)


# ============================================================
# 6. MODEL DEFINITION
# ============================================================

print("\n" + "=" * 70)
print("6. MODEL DEFINITION")
print("=" * 70)


models = {

    # --------------------------------------------------------
    # Linear Models
    # --------------------------------------------------------

    "Linear Regression":
        LinearRegression(),

    "Ridge":
        Ridge(
            alpha=1.0
        ),


    # --------------------------------------------------------
    # Decision Tree
    # --------------------------------------------------------

    "Decision Tree":
        DecisionTreeRegressor(

            random_state=RANDOM_STATE,

            max_depth=None,

            min_samples_leaf=2
        ),


    # --------------------------------------------------------
    # Random Forest
    # --------------------------------------------------------

    "Random Forest":
        RandomForestRegressor(

            n_estimators=300,

            max_features=1.0,

            min_samples_leaf=1,

            random_state=RANDOM_STATE,

            n_jobs=-1
        ),


    # --------------------------------------------------------
    # Gradient Boosting
    # --------------------------------------------------------

    "Gradient Boosting":
        GradientBoostingRegressor(

            n_estimators=200,

            learning_rate=0.03,

            max_depth=2,

            min_samples_leaf=2,

            random_state=RANDOM_STATE
        ),


    # --------------------------------------------------------
    # XGBoost
    # --------------------------------------------------------

    "XGBoost":
        XGBRegressor(

            n_estimators=300,

            learning_rate=0.03,

            max_depth=3,

            min_child_weight=1,

            subsample=0.8,

            colsample_bytree=0.8,

            reg_alpha=0.0,

            reg_lambda=1.0,

            objective="reg:squarederror",

            random_state=RANDOM_STATE,

            n_jobs=-1
        )
}


linear_models = [

    "Linear Regression",

    "Ridge"
]


tree_models = [

    "Decision Tree",

    "Random Forest",

    "Gradient Boosting",

    "XGBoost"
]


for model_name in models:

    print(
        f"- {model_name}"
    )

print("- DNN")


# ============================================================
# 7. DNN MODEL FUNCTION
# ============================================================

print("\n" + "=" * 70)
print("7. DNN MODEL")
print("=" * 70)


def build_dnn(input_dim):

    model = tf.keras.Sequential([

        tf.keras.layers.Input(
            shape=(input_dim,)
        ),

        tf.keras.layers.Dense(
            64,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            32,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            16,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            1,
            activation="linear"
        )

    ])


    model.compile(

        optimizer=tf.keras.optimizers.Adam(

            learning_rate=0.001

        ),

        loss="mse",

        metrics=[

            tf.keras.metrics.MeanAbsoluteError(

                name="mae"

            )

        ]

    )


    return model


print("DNN architecture:")
print("11 → 64 → 32 → 16 → 1")


# ============================================================
# 8. DNN CALLBACK FUNCTION
# ============================================================

def create_early_stopping():

    return tf.keras.callbacks.EarlyStopping(

        monitor="val_loss",

        patience=20,

        restore_best_weights=True,

        verbose=0

    )


# ============================================================
# 9. SKLEARN EVALUATION FUNCTION
# ============================================================

def evaluate_sklearn_model(

    model,

    X_train_data,

    y_train_data,

    X_val_data,

    y_val_data

):

    # --------------------------------------------------------
    # Clone model
    # --------------------------------------------------------

    model = clone(model)


    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    model.fit(

        X_train_data,

        y_train_data

    )


    # --------------------------------------------------------
    # Validation Prediction
    # --------------------------------------------------------

    predictions = model.predict(

        X_val_data

    )


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    mae = mean_absolute_error(

        y_val_data,

        predictions

    )


    rmse = np.sqrt(

        mean_squared_error(

            y_val_data,

            predictions

        )

    )


    r2 = r2_score(

        y_val_data,

        predictions

    )


    return (

        model,

        predictions,

        mae,

        rmse,

        r2

    )


# ============================================================
# 10. DNN EVALUATION FUNCTION
# ============================================================

def evaluate_dnn_model(

    X_train_data,

    y_train_data,

    X_val_data,

    y_val_data

):

    # --------------------------------------------------------
    # New DNN
    # --------------------------------------------------------

    model = build_dnn(

        input_dim=X_train_data.shape[1]

    )


    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    history = model.fit(

        X_train_data,

        y_train_data,

        validation_data=(

            X_val_data,

            y_val_data

        ),

        epochs=300,

        batch_size=32,

        callbacks=[

            create_early_stopping()

        ],

        verbose=0

    )


    # --------------------------------------------------------
    # Validation Prediction
    # --------------------------------------------------------

    predictions = model.predict(

        X_val_data,

        verbose=0

    ).reshape(-1)


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    mae = mean_absolute_error(

        y_val_data,

        predictions

    )


    rmse = np.sqrt(

        mean_squared_error(

            y_val_data,

            predictions

        )

    )


    r2 = r2_score(

        y_val_data,

        predictions

    )


    return (

        model,

        history,

        predictions,

        mae,

        rmse,

        r2

    )


# ============================================================
# 11. MODEL TRAINING + VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("11. MODEL TRAINING + VALIDATION")
print("=" * 70)


results = []

trained_models = {}

dnn_histories = {}

validation_predictions = {}


# ------------------------------------------------------------
# Dataset Loop
# ------------------------------------------------------------

for dataset_name in train_datasets.keys():

    print("\n")
    print("#" * 70)

    print(
        f"DATASET: {dataset_name}"
    )

    print("#" * 70)


    # --------------------------------------------------------
    # Raw dataset
    # --------------------------------------------------------

    X_train_raw, y_train_data = (

        train_datasets[dataset_name]

    )


    # --------------------------------------------------------
    # Scaled dataset
    # --------------------------------------------------------

    X_train_scaled_data, _ = (

        scaled_train_datasets[dataset_name]

    )


    # ========================================================
    # sklearn Models
    # ========================================================

    for model_name, model in models.items():

        print(
            f"\n[{dataset_name}] "
            f"{model_name}"
        )


        # ----------------------------------------------------
        # Linear Models
        # ----------------------------------------------------

        if model_name in linear_models:

            X_fit = X_train_scaled_data

            X_eval = X_val_scaled_df


        # ----------------------------------------------------
        # Tree / Ensemble Models
        # ----------------------------------------------------

        else:

            X_fit = X_train_raw

            X_eval = X_val


        try:

            (

                fitted_model,

                predictions,

                mae,

                rmse,

                r2

            ) = evaluate_sklearn_model(

                model,

                X_fit,

                y_train_data,

                X_eval,

                y_val

            )


            # ------------------------------------------------
            # Save model
            # ------------------------------------------------

            trained_models[

                (

                    dataset_name,

                    model_name

                )

            ] = fitted_model


            # ------------------------------------------------
            # Save predictions
            # ------------------------------------------------

            validation_predictions[

                (

                    dataset_name,

                    model_name

                )

            ] = predictions


            # ------------------------------------------------
            # Save results
            # ------------------------------------------------

            results.append({

                "Dataset":
                    dataset_name,

                "Model":
                    model_name,

                "MAE":
                    mae,

                "RMSE":
                    rmse,

                "R2":
                    r2

            })


            print(
                f"MAE  : {mae:.4f}"
            )

            print(
                f"RMSE : {rmse:.4f}"
            )

            print(
                f"R²   : {r2:.4f}"
            )


        except Exception as e:

            print(
                f"ERROR: {e}"
            )


    # ========================================================
    # DNN
    # ========================================================

    print(
        f"\n[{dataset_name}] DNN"
    )


    try:

        (

            dnn_model,

            history,

            predictions,

            mae,

            rmse,

            r2

        ) = evaluate_dnn_model(

            X_train_scaled_data,

            y_train_data,

            X_val_scaled_df,

            y_val

        )


        # ----------------------------------------------------
        # Save DNN model
        # ----------------------------------------------------

        trained_models[

            (

                dataset_name,

                "DNN"

            )

        ] = dnn_model


        # ----------------------------------------------------
        # Save history
        # ----------------------------------------------------

        dnn_histories[

            dataset_name

        ] = history


        # ----------------------------------------------------
        # Save predictions
        # ----------------------------------------------------

        validation_predictions[

            (

                dataset_name,

                "DNN"

            )

        ] = predictions


        # ----------------------------------------------------
        # Save result
        # ----------------------------------------------------

        results.append({

            "Dataset":
                dataset_name,

            "Model":
                "DNN",

            "MAE":
                mae,

            "RMSE":
                rmse,

            "R2":
                r2

        })


        print(
            f"MAE  : {mae:.4f}"
        )

        print(
            f"RMSE : {rmse:.4f}"
        )

        print(
            f"R²   : {r2:.4f}"
        )


    except Exception as e:

        print(
            f"DNN ERROR: {e}"
        )


# ============================================================
# 12. VALIDATION RESULTS
# ============================================================

print("\n" + "=" * 70)
print("12. VALIDATION RESULTS")
print("=" * 70)


validation_results = pd.DataFrame(

    results

)


validation_results = (

    validation_results

    .sort_values(

        by="RMSE",

        ascending=True

    )

    .reset_index(drop=True)

)


display(

    validation_results.round(4)

)


# ============================================================
# 13. RMSE RANKING
# ============================================================

print("\n" + "=" * 70)
print("13. RMSE RANKING")
print("=" * 70)


rmse_ranking = (

    validation_results

    .sort_values(

        "RMSE",

        ascending=True

    )

    .reset_index(drop=True)

)


display(

    rmse_ranking.round(4)

)


# ============================================================
# 14. MAE RANKING
# ============================================================

print("\n" + "=" * 70)
print("14. MAE RANKING")
print("=" * 70)


mae_ranking = (

    validation_results

    .sort_values(

        "MAE",

        ascending=True

    )

    .reset_index(drop=True)

)


display(

    mae_ranking.round(4)

)


# ============================================================
# 15. R² RANKING
# ============================================================

print("\n" + "=" * 70)
print("15. R² RANKING")
print("=" * 70)


r2_ranking = (

    validation_results

    .sort_values(

        "R2",

        ascending=False

    )

    .reset_index(drop=True)

)


display(

    r2_ranking.round(4)

)


# ============================================================
# 16. DATASET별 MODEL 비교
# ============================================================

print("\n" + "=" * 70)
print("16. DATASET별 MODEL 비교")
print("=" * 70)


for dataset_name in [

    "Original",

    "IQR Removed",

    "AE Removed"

]:

    print("\n" + "-" * 70)

    print(dataset_name)

    print("-" * 70)


    temp = (

        validation_results[

            validation_results["Dataset"]

            == dataset_name

        ]

        .sort_values(

            "RMSE"

        )

    )


    display(

        temp.round(4)

    )


# ============================================================
# 17. MODEL별 평균 성능
# ============================================================

print("\n" + "=" * 70)
print("17. MODEL별 평균 성능")
print("=" * 70)


model_summary = (

    validation_results

    .groupby("Model")

    [["MAE", "RMSE", "R2"]]

    .mean()

    .sort_values(

        "RMSE"

    )

)


display(

    model_summary.round(4)

)


# ============================================================
# 18. DATASET별 평균 성능
# ============================================================

print("\n" + "=" * 70)
print("18. DATASET별 평균 성능")
print("=" * 70)


dataset_summary = (

    validation_results

    .groupby("Dataset")

    [["MAE", "RMSE", "R2"]]

    .mean()

    .sort_values(

        "RMSE"

    )

)


display(

    dataset_summary.round(4)

)


# ============================================================
# 19. RMSE VISUALIZATION
# ============================================================

print("\n" + "=" * 70)
print("19. RMSE VISUALIZATION")
print("=" * 70)


plot_data = (

    validation_results

    .sort_values(

        "RMSE"

    )

)


labels = (

    plot_data["Dataset"]

    + " / "

    + plot_data["Model"]

)


plt.figure(

    figsize=(15, 7)

)


plt.bar(

    labels,

    plot_data["RMSE"]

)


plt.xlabel(

    "Dataset / Model"

)


plt.ylabel(

    "Validation RMSE"

)


plt.title(

    "Validation RMSE Comparison"

)


plt.xticks(

    rotation=75,

    ha="right"

)


plt.tight_layout()

plt.show()


# ============================================================
# 20. MAE VISUALIZATION
# ============================================================

print("\n" + "=" * 70)
print("20. MAE VISUALIZATION")
print("=" * 70)


plot_data = (

    validation_results

    .sort_values(

        "MAE"

    )

)


labels = (

    plot_data["Dataset"]

    + " / "

    + plot_data["Model"]

)


plt.figure(

    figsize=(15, 7)

)


plt.bar(

    labels,

    plot_data["MAE"]

)


plt.xlabel(

    "Dataset / Model"

)


plt.ylabel(

    "Validation MAE"

)


plt.title(

    "Validation MAE Comparison"

)


plt.xticks(

    rotation=75,

    ha="right"

)


plt.tight_layout()

plt.show()


# ============================================================
# 21. R² VISUALIZATION
# ============================================================

print("\n" + "=" * 70)
print("21. R² VISUALIZATION")
print("=" * 70)


plot_data = (

    validation_results

    .sort_values(

        "R2",

        ascending=False

    )

)


labels = (

    plot_data["Dataset"]

    + " / "

    + plot_data["Model"]

)


plt.figure(

    figsize=(15, 7)

)


plt.bar(

    labels,

    plot_data["R2"]

)


plt.xlabel(

    "Dataset / Model"

)


plt.ylabel(

    "Validation R²"

)


plt.title(

    "Validation R² Comparison"

)


plt.xticks(

    rotation=75,

    ha="right"

)


plt.tight_layout()

plt.show()


# ============================================================
# 22. DNN TRAINING CURVES
# ============================================================

print("\n" + "=" * 70)
print("22. DNN TRAINING CURVES")
print("=" * 70)


for dataset_name, history in dnn_histories.items():

    plt.figure(

        figsize=(8, 5)

    )


    plt.plot(

        history.history["loss"],

        label="Train Loss"

    )


    plt.plot(

        history.history["val_loss"],

        label="Validation Loss"

    )


    plt.xlabel(

        "Epoch"

    )


    plt.ylabel(

        "MSE Loss"

    )


    plt.title(

        f"DNN Training Curve - {dataset_name}"

    )


    plt.legend()


    plt.tight_layout()

    plt.show()


# ============================================================
# 23. SELECT FINAL CANDIDATE
# ============================================================
#
# IMPORTANT
# ------------------------------------------------------------
# Test data는 아직 사용하지 않는다.
#
# Validation RMSE가 가장 낮은 조합을
# 최종 후보로 선택한다.
#
# ============================================================

print("\n" + "=" * 70)
print("23. FINAL CANDIDATE SELECTION")
print("=" * 70)


best_row = (

    validation_results

    .sort_values(

        "RMSE",

        ascending=True

    )

    .iloc[0]

)


best_dataset = best_row["Dataset"]

best_model_name = best_row["Model"]


print(
    f"Dataset : {best_dataset}"
)

print(
    f"Model   : {best_model_name}"
)

print(
    f"MAE     : {best_row['MAE']:.4f}"
)

print(
    f"RMSE    : {best_row['RMSE']:.4f}"
)

print(
    f"R²      : {best_row['R2']:.4f}"
)


# ============================================================
# 24. FINAL TRAINING DATA PREPARATION
# ============================================================

print("\n" + "=" * 70)
print("24. FINAL MODEL PREPARATION")
print("=" * 70)


X_train_raw_final, y_train_final = (

    train_datasets[best_dataset]

)


X_train_scaled_final, _ = (

    scaled_train_datasets[best_dataset]

)


# ------------------------------------------------------------
# Linear / DNN
# ------------------------------------------------------------

if (

    best_model_name in linear_models

    or best_model_name == "DNN"

):

    X_fit_final = X_train_scaled_final

    X_test_final = X_test_scaled_df


# ------------------------------------------------------------
# Tree / Ensemble
# ------------------------------------------------------------

else:

    X_fit_final = X_train_raw_final

    X_test_final = X_test


print(
    "Final training shape:",
    X_fit_final.shape
)

print(
    "Final test shape:",
    X_test_final.shape
)


# ============================================================
# 25. FINAL MODEL TRAINING
# ============================================================

print("\n" + "=" * 70)
print("25. FINAL MODEL TRAINING")
print("=" * 70)


# ============================================================
# DNN
# ============================================================

if best_model_name == "DNN":

    final_model = build_dnn(

        input_dim=

        X_fit_final.shape[1]

    )


    final_history = final_model.fit(

        X_fit_final,

        y_train_final,

        validation_data=(

            X_val_scaled_df,

            y_val

        ),

        epochs=300,

        batch_size=32,

        callbacks=[

            create_early_stopping()

        ],

        verbose=0

    )


    test_predictions = (

        final_model

        .predict(

            X_test_final,

            verbose=0

        )

        .reshape(-1)

    )


# ============================================================
# sklearn Models
# ============================================================

else:

    final_model = clone(

        models[best_model_name]

    )


    final_model.fit(

        X_fit_final,

        y_train_final

    )


    test_predictions = (

        final_model

        .predict(

            X_test_final

        )

    )


# ============================================================
# 26. FINAL TEST EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("26. FINAL TEST EVALUATION")
print("=" * 70)


test_mae = mean_absolute_error(

    y_test,

    test_predictions

)


test_rmse = np.sqrt(

    mean_squared_error(

        y_test,

        test_predictions

    )

)


test_r2 = r2_score(

    y_test,

    test_predictions

)


print(
    f"Dataset : {best_dataset}"
)

print(
    f"Model   : {best_model_name}"
)

print()

print(
    f"MAE     : {test_mae:.4f}"
)

print(
    f"RMSE    : {test_rmse:.4f}"
)

print(
    f"R²      : {test_r2:.4f}"
)


# ============================================================
# 27. VALIDATION vs TEST
# ============================================================

print("\n" + "=" * 70)
print("27. VALIDATION vs TEST")
print("=" * 70)


validation_best = validation_results[

    (

        validation_results["Dataset"]

        == best_dataset

    )

    &

    (

        validation_results["Model"]

        == best_model_name

    )

].iloc[0]


final_comparison = pd.DataFrame({

    "Metric": [

        "MAE",

        "RMSE",

        "R²"

    ],

    "Validation": [

        validation_best["MAE"],

        validation_best["RMSE"],

        validation_best["R2"]

    ],

    "Test": [

        test_mae,

        test_rmse,

        test_r2

    ]

})


display(

    final_comparison.round(4)

)


# ============================================================
# 28. ACTUAL vs PREDICTED
# ============================================================

print("\n" + "=" * 70)
print("28. ACTUAL vs PREDICTED")
print("=" * 70)


plt.figure(

    figsize=(7, 7)

)


plt.scatter(

    y_test,

    test_predictions,

    alpha=0.6

)


min_value = min(

    y_test.min(),

    test_predictions.min()

)


max_value = max(

    y_test.max(),

    test_predictions.max()

)


plt.plot(

    [min_value, max_value],

    [min_value, max_value],

    linestyle="--"

)


plt.xlabel(

    "Actual Quality"

)


plt.ylabel(

    "Predicted Quality"

)


plt.title(

    f"Actual vs Predicted\n"

    f"{best_model_name} / {best_dataset}"

)


plt.tight_layout()

plt.show()


# ============================================================
# 29. RESIDUAL ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("29. RESIDUAL ANALYSIS")
print("=" * 70)


residuals = (

    y_test.values

    - test_predictions

)


prediction_result = pd.DataFrame({

    "Actual":

        y_test.values,

    "Predicted":

        test_predictions,

    "Residual":

        residuals

})


prediction_result[

    "Absolute Error"

] = (

    prediction_result["Residual"]

    .abs()

)


print(
    "Largest prediction errors:"
)


display(

    prediction_result

    .sort_values(

        "Absolute Error",

        ascending=False

    )

    .head(20)

    .round(4)

)


# ============================================================
# 30. RESIDUAL DISTRIBUTION
# ============================================================

plt.figure(

    figsize=(8, 5)

)


plt.hist(

    residuals,

    bins=30

)


plt.axvline(

    0,

    linestyle="--"

)


plt.xlabel(

    "Residual"

)


plt.ylabel(

    "Frequency"

)


plt.title(

    "Residual Distribution"

)


plt.tight_layout()

plt.show()


# ============================================================
# 31. RESIDUAL vs PREDICTED
# ============================================================

plt.figure(

    figsize=(8, 5)

)


plt.scatter(

    test_predictions,

    residuals,

    alpha=0.6

)


plt.axhline(

    0,

    linestyle="--"

)


plt.xlabel(

    "Predicted Quality"

)


plt.ylabel(

    "Residual"

)


plt.title(

    "Residual vs Predicted"

)


plt.tight_layout()

plt.show()


# ============================================================
# 32. FEATURE IMPORTANCE
# ============================================================
#
# Tree-based model인 경우에만 실행
#
# ============================================================

print("\n" + "=" * 70)
print("32. FEATURE IMPORTANCE")
print("=" * 70)


if hasattr(

    final_model,

    "feature_importances_"

):

    feature_importance = pd.DataFrame({

        "Feature":

            X_train.columns,

        "Importance":

            final_model.feature_importances_

    })


    feature_importance = (

        feature_importance

        .sort_values(

            "Importance",

            ascending=False

        )

        .reset_index(drop=True)

    )


    display(

        feature_importance.round(4)

    )


    plt.figure(

        figsize=(9, 6)

    )


    plt.barh(

        feature_importance["Feature"],

        feature_importance["Importance"]

    )


    plt.xlabel(

        "Importance"

    )


    plt.ylabel(

        "Feature"

    )


    plt.title(

        f"Feature Importance\n"

        f"{best_model_name}"

    )


    plt.gca().invert_yaxis()


    plt.tight_layout()

    plt.show()


else:

    print(

        "Feature importance is not directly available "

        "for this model."

    )


# ============================================================
# 33. SAVE RESULTS
# ============================================================

print("\n" + "=" * 70)
print("33. SAVE RESULTS")
print("=" * 70)


validation_results.to_csv(

    "wine_validation_results.csv",

    index=False

)


final_comparison.to_csv(

    "wine_final_test_results.csv",

    index=False

)


prediction_result.to_csv(

    "wine_test_predictions.csv",

    index=False

)


dataset_info_df.to_csv(

    "wine_modeling_dataset_info.csv",

    index=False

)


if (

    "feature_importance"

    in globals()

):

    feature_importance.to_csv(

        "wine_feature_importance.csv",

        index=False

    )


print(
    "Validation results     → wine_validation_results.csv"
)

print(
    "Final test results     → wine_final_test_results.csv"
)

print(
    "Test predictions       → wine_test_predictions.csv"
)

print(
    "Dataset information    → wine_modeling_dataset_info.csv"
)


if (

    "feature_importance"

    in globals()

):

    print(
        "Feature importance     → wine_feature_importance.csv"
    )


# ============================================================
# 34. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("MODELING COMPLETE")
print("=" * 70)


print(
    f"Final Dataset : {best_dataset}"
)

print(
    f"Final Model   : {best_model_name}"
)

print()

print(
    f"Test MAE      : {test_mae:.4f}"
)

print(
    f"Test RMSE     : {test_rmse:.4f}"
)

print(
    f"Test R²       : {test_r2:.4f}"
)

print()

print(
    "Validation was used for model selection."
)

print(
    "Test was used only for final evaluation."
)

print("=" * 70)
```
